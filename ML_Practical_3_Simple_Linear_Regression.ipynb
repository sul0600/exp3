# Practical 3: Simple Linear Regression

**Aim:** To predict employee salary based on years of experience using Simple Linear Regression.

**Dataset:** Salary Data (Kaggle)

**Kaggle link:** https://www.kaggle.com/datasets/karthickveerakumar/salary-data-simple-linear-regression

### Dataset setup
1. Download the dataset from Kaggle.
2. Extract the ZIP file.
3. Keep `Salary_Data.csv` in the same folder as this notebook.
4. Run the cells one by one using Shift + Enter.

This notebook follows your uploaded `exp3.ipynb` style: load X and y, plot the data, create `LinearRegression()`, use `fit()`, `predict()`, `coef_`, `intercept_` and `score()`.

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## 2. Load the dataset

```python
df = pd.read_csv("Salary_Data.csv")
df.head()
```

## 3. Display the dataset

```python
df.info()
```

```python
df.describe()
```

## 4. Separate X and y

```python
X = df[["YearsExperience"]]
y = df["Salary"]

print(X.head())
print(y.head())
```

## 5. Display the data using scatter plot

```python
plt.title("Salary vs Years of Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.scatter(X, y, color="red")
plt.show()
```

## 6. Initialize the algorithm

```python
model = LinearRegression()
```

## 7. Train the model using fit()

```python
model.fit(X, y)
```

## 8. Predict salary using predict()

```python
# Predict salary for an employee with 5 years of experience
ip = [[5]]
model.predict(ip)
```

## 9. Display coefficient, intercept and accuracy

```python
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
print("Accuracy:", model.score(X, y) * 100, "%")
```

## 10. Display the regression line

```python
plt.title("Simple Linear Regression")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.scatter(X, y, color="red")
plt.plot(X, model.predict(X), color="blue")
plt.show()
```

## Conclusion

Thus, Simple Linear Regression was implemented to predict employee salary based on years of experience. The model was trained using `fit()` and salary was predicted using `predict()`. The regression line shows the relationship between experience and salary.
